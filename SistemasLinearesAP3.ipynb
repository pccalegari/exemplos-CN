**Problema 2 - Refinamento de Solução**

Ao resolver um sistema linear $Ax=b$ por um método direto podemos obter uma solução aproximada $x_0$, devido aos erros de arredondamento que podem resultar em uma solução ruim. Vamos definir o resíduo associado a solução $x_0$, $r = b - Ax_0$ e o erro $e = x - x_0$. Dessa forma, podemos definir o problema residual $Ae = r$, já que $Ax = b \Longrightarrow A(x_0 + e) = b \Longrightarrow Ae = b - Ax_0 = r.$

É razoável supor que $x_1 = x_0 + e$ será uma solução de $Ax = b$ melhor que $x_0$. Assim construimos uma sequência de aproximações $Ae=r_i$, com $r_i = b - Ax_i$ e $x_{i+1} = x_{i} + e$.
