# Unidade 3 - Sistemas de equações lineares

**Métodos para sistemas triangulares**

Um sistema linear triangular superior pode ser resolvido com o método da substituição inversa. Admitindo-se que os elementos da diagonal são não nulos, ou seja, $a_{ii}\neq 0, \forall i$, o método consiste em resolver a última equação, depois substituir o resultado na penúltima e resolvê-la e assim por diante. Já um sistema triangular inferior pode ser resolvido usando-se o método da substituição direta.

**Método da substituição inversa**

Dados de entrada: $A$ uma matriz triangular superior com $a_{ii}\neq 0,\forall i$ e $a_{ij}=0$, se $i>j$ e $b_i$ com $1 < i,j < n$.

Saída: uma aproximação para $x$ tal que $Ax = b$.

```
1.  x[n] = b[n]/A[n][n]$
2.  soma = 0
3.  Para k = n-1 até 1
4.     soma = b[k]
5.     Para j = k+1 até n
6.        soma = soma - A[k][j]x[j]
7.     x[k] = soma/A[k][k]

```
O número de operações que o método realiza é de aproximadamente $n^2$, ou seja, $O(n^2)$. Atenção com os índices ao implementar os algoritmos dessa unidade.

**Método da substituição direta**

Dados de entrada: $A$ uma matriz triangular inferior com $a_{ii}\neq 0,\forall i$ e $a_{ij}=0$, se $i < j$ e $b_i$ com $1 < i,j < n$.

Saída: uma aproximação para $x$ tal que $Ax = b$.

```
1.  x[1]=b[1]/A[1][1]$
2.  soma = 0
3.  Para k = 2 até n
4.     soma = b[k]
5.     Para j = 1 até k-1
6.        soma = soma - A[k][j]x[j]
7.     x[k] = soma/A[k][k]

```

**Método de Eliminação de Gauss**

É um método direto que consiste em transformar o sistema de euações lineares em sistema de equações lineares triangular superior. Essa transformação é feita por meio de operações elementares que não alteram a solução do sistema.

**Como transformar um sistema linear em um sistema linear triangular superior equivalente:**

As operações elementares sobre as linhas para transformar um sistema linear em um sistema linear triangular superior equivalente são:


1.   Multiplicar uma equação por uma constante não nula.
2.   Somar uma equação multiplicada por uma constante a outra equação.
3.   Trocar a ordem das equações.

Na forma matricial, o sistema de equações é dado por:

$\displaystyle{\left(\begin{array}{ccccc}
a_{11} & a_{12} & a_{13} & \ldots & a_{1n} \\
a_{21} & a_{22} & a_{23} & \ldots & a_{2n} \\
 \vdots & \vdots & \vdots & \ldots & \vdots \\
 a_{n1} & a_{n2} & a_{n3} & \ldots & a_{nn}\\
\end{array}\right )\left(\begin{array}{c}
x_1\\
x_2\\
\vdots\\
x_n\\
\end{array}\right )=\left(\begin{array}{c}
b_1\\
b_2\\
\vdots\\
b_n\\
\end{array}\right )}$

O nosso objetivo é transformar o sistema em um sistema triangular equivalente:


$\displaystyle{\left(\begin{array}{ccccc}
a_{11} & a_{12} & a_{13} & \ldots & a_{1n} \\
0 & a_{22}^{(1)} & a_{23}^{(1)} & \ldots & a_{2n}^{(1)} \\
 \vdots & \vdots & \vdots & \ldots & \vdots \\
0 & 0 & 0 & \ldots & a_{nn}^{(n-1)}\\
\end{array}\right )\left(\begin{array}{c}
x_1\\
x_2\\
\vdots\\
x_n\\
\end{array}\right )=\left(\begin{array}{c}
b_1\\
b_2^{(1)}\\
\vdots\\
b_n^{(n-1)}\\
\end{array}\right )}$

*Passo 1*. Anular todos os elementos da coluna 1 da matriz $A$ abaixo da diagonal principal. Para isso faremos: $L_k \leftarrow L_k - m_{k1}L_1,$ onde $L_k$ é a $k$-ésima linha, com $2\le k\le n$ e $m_{k1}$ é o multiplicador dado por $m_{k1}=\frac{a_{k1}}{a_{11}}$.

*Passo 2*. Na matriz alterada pelo passo 1, anular os elementos da coluna 2, abaixo da diagonal principal. Para isso, faremos: $L_k\leftarrow L_k - m_{k2}L_2,$ com $3\le k\le n$ e $m_{k2}=\frac{a^{(1)}_{k2}}{a^{(1)}_{22}}$.

Repetindo o processo até a penúltima coluna ($n-1$) obtemos o sistema triangular superior equivalente. Em seguida, resolvemos o sistema equivalente por meio do método da substituição inversa.

O tipo de erro envolvido neste método é o erro de arredondamento, que pode ser desastroso.
